```python
df1 = df.rename(columns={'price 2': 'target'})
df1_clean = df1.dropna()

y = df1_clean['target']
X = df1_clean.drop(
    columns=['target', 'price', 'page 2 (clothing model)', 'session ID', 'year']
)

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=40, stratify=y
)

from imblearn.over_sampling import SMOTE

smote = SMOTE(random_state=42)
X_train_smote, y_train_smote = smote.fit_resample(X_train, y_train)

print("SMOTE resampling completed successfully!")
print("X_train shape:", X_train_smote.shape)
print("y_train value counts:\n", y_train_smote.value_counts())
```

```python
from sklearn.feature_selection import SelectKBest, f_classif

skb = SelectKBest(score_func=f_classif, k=8)

X_train_selected = skb.fit_transform(X_train_smote, y_train_smote)
X_test_selected = skb.transform(X_test)

print("Training data shape:", X_train_selected.shape)
print("Testing data shape:", X_test_selected.shape)
```

```python
from sklearn.preprocessing import PowerTransformer
import pandas as pd

selected_columns = X.columns[skb.get_support()]

pt = PowerTransformer(method='yeo-johnson')

X_train_transformed = pt.fit_transform(X_train_selected)

X_test_transformed = pt.transform(X_test_selected)

X_train_transformed = pd.DataFrame(
    X_train_transformed,
    columns=selected_columns
)

X_test_transformed = pd.DataFrame(
    X_test_transformed,
    columns=selected_columns
)

print("Training transformed shape:", X_train_transformed.shape)
print("Testing transformed shape:", X_test_transformed.shape)

X_train_transformed.head()
```

```python
from sklearn.preprocessing import StandardScaler

ss = StandardScaler()

X_train_scaled = ss.fit_transform(X_train_transformed)
X_test_scaled = ss.transform(X_test_transformed)

print("Training scaled shape:", X_train_scaled.shape)
print("Testing scaled shape:", X_test_scaled.shape)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
```

```python
LR = LogisticRegression(max_iter=1000)

LR.fit(X_train, y_train)

y_pred_lr = LR.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, f1_score, precision_score, recall_score

accuracy_lr = accuracy_score(y_test, y_pred_lr)
precision_lr = precision_score(y_test, y_pred_lr, average='weighted')
recall_lr = recall_score(y_test, y_pred_lr, average='weighted')
f1_lr = f1_score(y_test, y_pred_lr, average='weighted')

print("Logistic Regression")
print("Accuracy:", accuracy_lr)
print("Precision:", precision_lr)
print("Recall:", recall_lr)
print("F1 Score:", f1_lr)
```

```python
DT = DecisionTreeClassifier(random_state=40)

DT.fit(X_train, y_train)

y_pred_dt = DT.predict(X_test)
```

```python
accuracy_dt = accuracy_score(y_test, y_pred_dt)
precision_dt = precision_score(y_test, y_pred_dt, average='weighted')
recall_dt = recall_score(y_test, y_pred_dt, average='weighted')
f1_dt = f1_score(y_test, y_pred_dt, average='weighted')

print("Decision Tree Classifier")
print("Accuracy:", accuracy_dt)
print("Precision:", precision_dt)
print("Recall:", recall_dt)
print("F1 Score:", f1_dt)
```

```python
RF = RandomForestClassifier(
    n_estimators=100,
    random_state=40
)

RF.fit(X_train, y_train)

y_pred_rf = RF.predict(X_test)
```

```python
accuracy_rf = accuracy_score(y_test, y_pred_rf)
precision_rf = precision_score(y_test, y_pred_rf, average='weighted')
recall_rf = recall_score(y_test, y_pred_rf, average='weighted')
f1_rf = f1_score(y_test, y_pred_rf, average='weighted')

print("Random Forest Classifier")
print("Accuracy:", accuracy_rf)
print("Precision:", precision_rf)
print("Recall:", recall_rf)
print("F1 Score:", f1_rf)
```

```python
comparison = pd.DataFrame({
    'Model': [
        'Logistic Regression',
        'Decision Tree',
        'Random Forest'
    ],
    
    'Accuracy': [
        accuracy_lr,
        accuracy_dt,
        accuracy_rf
    ],
    
    'Precision': [
        precision_lr,
        precision_dt,
        precision_rf
    ],
    
    'Recall': [
        recall_lr,
        recall_dt,
        recall_rf
    ],
    
    'F1 Score': [
        f1_lr,
        f1_dt,
        f1_rf
    ]
})

comparison
```
